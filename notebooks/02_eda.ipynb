## 1. Data Quality Callout Section

### The Pre-Registration Sales Anomaly
> **Issue Summary:** Approximately **19.92%** of sales records (996 out of 5000 transactions) have a transaction `Date` that is strictly earlier than the customer's recorded `JoinDate` in the raw data. This represents a registration-date data anomaly, which has been flagged in the cleaned sales table as `pre_registration_flag = 1`.
>
> **Handling Strategy:**
> - **Sales, Product, and Regional Analysis:** We use all 5000 transactions (regardless of the flag) to ensure total revenue and sales volumes are accurate.
> - **Customer Lifecycle and Tenure-based Analysis:** We exclude the 5 customers with negative tenure to prevent distorted tenure averages or incorrect lifecycle classifications. This ensures that tenure calculations are strictly positive and based on logical sign-up dates.

## 8. Summary of Business Insights

Here are 5 core business insights generated from this exploratory analysis to carry forward:

1. **Strong Customer Retention:** **83.8% of transactions** are placed by returning customers (orders placed >90 days after registration), showing high brand loyalty and strong post-acquisition engagement.
2. **Discount Inefficiency:** Offering higher discounts (10% and 15%) dramatically decreases average profit margins but **fails to drive any significant increase in sales volume**. Setting a standard flat-discount ceiling of 5% or implementing loyalty-only rewards could recover significant lost profits.
3. **High-Value Product Drivers:** High-ticket categories (Electronics, specifically Camera and TV subcategories) maintain the highest average profit margins and drive gross revenue, whereas lower-cost Apparel/Accessories drive volume. Bundling accessories with high-value purchases presents a clear cross-selling opportunity.
4. **Geographical Concentration:** The **South and East regions** are the dominant revenue and profit centers, led by stores in Jimenezborough and Johnmouth. The West and North regions are currently underperforming, representing either untapped potential or operational bottlenecks.
5. **Summer Seasonality Pattern:** Sales volume and revenue experience consistent surges in June and July across both 2024 and 2025. Inventory levels and marketing spend should be scaled up in late Q2 to capitalize on this repeatable seasonal peak.
